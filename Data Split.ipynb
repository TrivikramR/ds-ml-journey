```python
# Basic liberaries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# sklearn imports
from sklearn.model_selection import train_test_split
# Ignore warnings

import warnings
warnings.filterwarnings('ignore')
```

```python
data = pd.read_csv('Housing.csv')
data.head()
```

```python
encoded_data = pd.get_dummies(data, dtype='int',drop_first=True)
encoded_data.head()
```

```python
encoded_data.columns = encoded_data.columns.str.replace('_yes', '')
encoded_data.columns = encoded_data.columns.str.replace('furnishingstatus_','')
encoded_data.head()
```

```python
# Test train split

x = encoded_data.drop(['price'],axis=1)
y = encoded_data[['price']]

#train-test-split

x_train,x_test,y_train,y_test = train_test_split(x, y, test_size=0.3, random_state=42)
print('x_train dimension:',x_train.shape)
print('x_test dimension:',x_test.shape)
print('y_train dimension:',y_train.shape)
print('y_test dimension:',y_test.shape)
```

```python
penguin_data = sns.load_dataset('penguins')
penguin_data.head()
```

```python
penguin_data.dropna(inplace=True)
```

```python
encoded_pdata = pd.get_dummies(penguin_data, dtype='int',drop_first=True)
encoded_pdata.head()
```

```python
encoded_pdata.head()
```

```python
x = encoded_pdata.drop(['body_mass_g'],axis=1)
y = encoded_pdata[['body_mass_g']]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42)
print('x_train dimension:',x_train.shape)
print('x_test dimension:',x_test.shape)
print('y_train dimension:',y_train.shape)
print('y_test dimension:',y_test.shape)
```
