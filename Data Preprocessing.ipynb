```python
# Basic liberaries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# sklearn imports

from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import train
# Ignore warnings

import warnings
warnings.filterwarnings('ignore')
```

```python
data = pd.read_csv('Housing.csv')
data.head()
```

```python
data.isnull().sum()
```

```python
data.dtypes
```

```python
# filtering the cat columns

# cat_col=[]

# for cols in data.columns:
#     if data[cols].dtype =='O':
#         cat_cols.append(cols)
# print(cat_cols)

# List comprehension to find the list of categorical columns

cat_cols = [cols for cols in data.columns if data[cols].dtype=='O']
cat_cols
```

```python
# unique values for categorical columns

for cols in cat_cols:
    print(cols,':',data[cols].unique())        
    
```

#  OneHotEncoder

```python
# Using OneHotEncoder

sk_encoder = OneHotEncoder()
sk_encoded_data = sk_encoder.fit_transform(data[cat_cols])
sk_encoded_data
```

```python
# the above data is in compresed format
# unzip it
sk_encoder = OneHotEncoder(sparse_output=False)
sk_encoded_data = sk_encoder.fit_transform(data[cat_cols])
sk_encoded_data
```

```python
# the return type above is array
# lets convert to DF
sk_encoder = OneHotEncoder(sparse_output=False)
sk_encoded_data = sk_encoder.fit_transform(data[cat_cols])
sk_encoded_data = pd.DataFrame(sk_encoded_data)
sk_encoded_data
```

```python
# from the above we are not getting columns names.
# lets get the column names

sk_encoder = OneHotEncoder(sparse_output=False)
sk_encoded_data = sk_encoder.fit_transform(data[cat_cols])
sk_encoded_data = pd.DataFrame(sk_encoded_data, columns=sk_encoder.get_feature_names_out())
sk_encoded_data.head()
```

```python
# from above we are getting yes and no
# we will drop the column (yes or no)

sk_encoder = OneHotEncoder(sparse_output=False, drop='first')
sk_encoded_data = sk_encoder.fit_transform(data[cat_cols])
sk_encoded_data = pd.DataFrame(sk_encoded_data, columns=sk_encoder.get_feature_names_out())
sk_encoded_data.head()
```

```python
# Every value is float
# let convert it to int

sk_encoder = OneHotEncoder(sparse_output=False, drop='first', dtype='int')
sk_encoded_data = sk_encoder.fit_transform(data[cat_cols])
sk_encoded_data = pd.DataFrame(sk_encoded_data, columns=sk_encoder.get_feature_names_out())
sk_encoded_data.head()
```

```python
# we only got cat data.
# we need to merge the cat and num data

sk_encoder = OneHotEncoder(sparse_output=False, drop='first', dtype='int')
sk_encoded_data = sk_encoder.fit_transform(data[cat_cols])
sk_encoded_data = pd.DataFrame(sk_encoded_data, columns=sk_encoder.get_feature_names_out())
sk_encoded_data = pd.concat([data.drop(cat_cols, axis=1),sk_encoded_data],axis=1)
sk_encoded_data.head()
```

# OneHotEncoder 2nd Method

```python
# Pandas - get dummies

getdummies_df = pd.get_dummies(data, dtype='int', drop_first=True)
getdummies_df.head()
```

#  Observation

* Use OneHotEncoder for large DF
* Use get dummies for small DF (less than ~ 100000)

# Another example for OneHotEncoder

```python
penguin_data = sns.load_dataset('penguins')
penguin_data.head()
```

```python
pcat_cols = [cols for cols in penguin_data.columns if penguin_data[cols].dtype =='O' ]
pcat_cols
```

```python
for cols in pcat_cols:
    print(cols,':',penguin_data[cols].unique())
```

```python
sk_encoder = OneHotEncoder(sparse_output=False, dtype='int',drop='first')
sk_encoded_data = sk_encoder.fit_transform(penguin_data[pcat_cols])
sk_encoded_data = pd.DataFrame(sk_encoded_data, columns=sk_encoder.get_feature_names_out())
sk_encoded_data = pd.concat([penguin_data.drop(pcat_cols, axis=1),sk_encoded_data],axis=1)
sk_encoded_data.head()
```

```python
getdummies_df = pd.get_dummies(penguin_data, dtype='int', drop_first=True)
getdummies_df.head()
```
