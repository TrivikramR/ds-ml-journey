```python
# Basic liberaries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# sklearn liberaries
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler , OneHotEncoder
# Ignore warnings

import warnings
warnings.filterwarnings('ignore')
```

```python
# Importing the data
data = pd.read_csv('Housing.csv')
data.head()
```

```python
# Shape of the data
data.shape
```

```python
# missing values
data.isnull().sum()
```

```python
# Dependency split

x = data.drop('price',axis=1)
y = data[['price']]

print(x.shape)
print(y.shape)
```

```python
# Train Test Split on x, y

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42)
print(x_train.shape,x_test.shape,y_train.shape,y_test.shape, sep='\n')
```

```python
# Filtering the categorical columns
cat_cols = [cols for cols in data.columns if data[cols].dtype == 'O']
```

```python
# OneHotEncoding on train test split data
# fit transform on train data

encoder = OneHotEncoder(sparse_output=False, dtype='int', drop='first')
x_train_encoded = encoder.fit_transform(x_train[cat_cols])
x_train_encoded = pd.DataFrame(x_train_encoded, columns = encoder.get_feature_names_out())
x_train_encoded_df = pd .concat([x_train.drop(cat_cols, axis=1), x_train_encoded],axis=1)
x_train_encoded_df.head()
```

```python
x_train_encoded_df.shape
```

# Observation

* if you can see the shape of above train data, the shape has changed from 381 to 491.
* it is because of the index's when we are concatenating both x_train.drop(cat_cols, axis=1), x_train_encoded],axis=1.
* As we are concatenating the new DF is reseting the index's.

```python
#Resetting the index use index=

encoder = OneHotEncoder(sparse_output=False, dtype='int', drop='first')
x_train_encoded = encoder.fit_transform(x_train[cat_cols])
x_train_encoded = pd.DataFrame(x_train_encoded,index=x_train.index, columns = encoder.get_feature_names_out())
x_train_encoded_df = pd .concat([x_train.drop(cat_cols, axis=1), x_train_encoded],axis=1)
x_train_encoded_df.head()
```

```python
x_train_encoded_df.shape
```

```python
# test data
#Transform on test

x_test_encoded = encoder.transform(x_test[cat_cols])
x_test_encoded = pd.DataFrame(x_test_encoded, index=x_test.index, columns = encoder.get_feature_names_out())
x_test_encoded_df = pd .concat([x_test.drop(cat_cols, axis=1), x_test_encoded],axis=1)
x_test_encoded_df.head()
```

```python
x_test_encoded_df.shape
```

```python
# Scaling

scaler = StandardScaler()

#fit transform on train

x_train_scaled = scaler.fit_transform(x_train_encoded_df)
x_train_scaled_df = pd.DataFrame(x_train_scaled, index=x_train_encoded_df.index, columns=scaler.get_feature_names_out())

#transform on test

x_test_scaled = scaler.transform(x_test_encoded_df)
x_test_scaled_df = pd.DataFrame(x_test_scaled, index=x_test_encoded_df.index, columns=scaler.get_feature_names_out())
```
