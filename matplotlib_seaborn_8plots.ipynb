## Setup

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Global style — run this once at the top of every notebook
sns.set_theme(style='whitegrid')
plt.rcParams['figure.dpi'] = 150
plt.rcParams['font.size'] = 12

# Load and clean data
df = pd.read_csv('diabetes.csv')  # update path if needed
df_clean = df.copy()
df_clean['Insulin_Missing'] = (df_clean['Insulin'] == 0).astype(int)
df_clean['SkinThickness_Missing'] = (df_clean['SkinThickness'] == 0).astype(int)
cols = ['Glucose', 'BloodPressure', 'BMI', 'Insulin', 'SkinThickness']
df_clean[cols] = df_clean[cols].replace(0, np.nan)
for col in cols:
    df_clean[col].fillna(df_clean[col].median(), inplace=True)

print('Ready:', df_clean.shape)
df_clean.head()
```

---
## Plot 1 — Histogram
**When to use:** Distribution of one numerical column  
**When NOT to use:** Categorical columns — use countplot instead

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))

# Basic histogram
axes[0].hist(df_clean['Glucose'], bins=30, 
             color='steelblue', edgecolor='black')
axes[0].set_title('Basic — Matplotlib')
axes[0].set_xlabel('Glucose')
axes[0].set_ylabel('Count')

# With KDE curve
sns.histplot(df_clean['Glucose'], bins=30, 
             kde=True, ax=axes[1], color='steelblue')
axes[1].set_title('With KDE Curve — Seaborn')
axes[1].set_xlabel('Glucose')

# Split by target — most useful version
sns.histplot(data=df_clean, x='Glucose', hue='Outcome',
             bins=30, kde=True, ax=axes[2], alpha=0.6,
             palette=['steelblue', 'salmon'])
axes[2].set_title('Split by Target — Most Useful')
axes[2].set_xlabel('Glucose')

plt.suptitle('Plot 1: Histogram', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

**Observation:** *(what do you see in the glucose distribution?)*

---
## Plot 2 — Boxplot
**When to use:** Spotting outliers, comparing distributions across categories  
**When NOT to use:** When you care about exact distribution shape — use histogram

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))

# Single column
sns.boxplot(data=df_clean, y='BMI', ax=axes[0], color='steelblue')
axes[0].set_title('Single Column')

# vs Target
sns.boxplot(data=df_clean, x='Outcome', y='BMI',
            palette=['steelblue', 'salmon'], ax=axes[1])
axes[1].set_title('vs Target — Most Useful')
axes[1].set_xticklabels(['No Diabetes', 'Diabetes'])

# All columns at once
original_cols = ['Glucose', 'BMI', 'Age', 'Pregnancies',
                  'Insulin', 'BloodPressure',
                  'SkinThickness', 'DiabetesPedigreeFunction']
df_clean[original_cols].boxplot(ax=axes[2])
axes[2].set_title('All Columns at Once')
axes[2].tick_params(axis='x', rotation=45)

plt.suptitle('Plot 2: Boxplot', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

**Observation:** *(which column has the most outliers? does BMI differ between classes?)*

---
## Plot 3 — Correlation Heatmap
**When to use:** Understanding relationships between ALL features at once  
**When NOT to use:** Categorical columns — encode them first

```python
plt.figure(figsize=(12, 8))
sns.heatmap(
    df_clean.corr(),
    annot=True,        # show correlation numbers
    fmt='.2f',         # 2 decimal places
    cmap='coolwarm',   # red=positive, blue=negative
    center=0,          # white at zero correlation
    square=True,       # square cells
    linewidths=0.5     # grid lines between cells
)
plt.title('Plot 3: Correlation Heatmap — Diabetes Dataset',
          fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# Sorted correlation with target
print('Correlation with Outcome (sorted):')
print(df_clean.corr()['Outcome'].sort_values(ascending=False).round(3))
```

**Observation:** *(which two non-target features correlate most with each other?)*

---
## Plot 4 — Countplot
**When to use:** Distribution of categorical columns, class balance  
**When NOT to use:** Numerical columns — use histogram

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Simple count
sns.countplot(data=df_clean, x='Outcome',
              palette=['steelblue', 'salmon'], ax=axes[0])
axes[0].set_title('Target Class Distribution')
axes[0].set_xticklabels(['No Diabetes', 'Diabetes'])
axes[0].set_xlabel('')

# Add percentage labels on bars
total = len(df_clean)
for p in axes[0].patches:
    percentage = f'{100 * p.get_height() / total:.1f}%'
    axes[0].annotate(percentage,
                     (p.get_x() + p.get_width() / 2, p.get_height()),
                     ha='center', va='bottom', fontsize=12)

# Pregnancies distribution by outcome
sns.countplot(data=df_clean, x='Pregnancies',
              hue='Outcome',
              palette=['steelblue', 'salmon'], ax=axes[1])
axes[1].set_title('Pregnancies Count by Outcome')
axes[1].legend(['No Diabetes', 'Diabetes'])

plt.suptitle('Plot 4: Countplot', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

**Observation:** *(is the dataset balanced? what do you notice about pregnancies?)*

---
## Plot 5 — Scatterplot
**When to use:** Relationship between two numerical columns  
**When NOT to use:** More than 2 variables — use pairplot instead

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Basic scatterplot
sns.scatterplot(data=df_clean, x='Glucose', y='BMI',
                ax=axes[0], alpha=0.5, color='steelblue')
axes[0].set_title('Glucose vs BMI — Basic')

# Colored by target — most useful version
sns.scatterplot(data=df_clean, x='Glucose', y='BMI',
                hue='Outcome', alpha=0.6, ax=axes[1],
                palette=['steelblue', 'salmon'])
axes[1].set_title('Glucose vs BMI — Colored by Outcome')
axes[1].legend(['No Diabetes', 'Diabetes'])

plt.suptitle('Plot 5: Scatterplot', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

**Observation:** *(do diabetic and non-diabetic patients cluster separately or overlap?)*

---
## Plot 6 — Pairplot
**When to use:** Quick overview of ALL pairwise relationships  
**When NOT to use:** More than 6 features — too slow and unreadable

```python
# Always select key features only — never all columns
key_features = ['Glucose', 'BMI', 'Age', 'Insulin', 'Outcome']

sns.pairplot(
    df_clean[key_features],
    hue='Outcome',
    diag_kind='kde',          # distribution on diagonal
    plot_kws={'alpha': 0.5},  # transparency
    palette=['steelblue', 'salmon']
)
plt.suptitle('Plot 6: Pairplot — Key Diabetes Features',
             y=1.02, fontsize=14, fontweight='bold')
plt.show()
```

**Observation:** *(which pair of features shows the clearest class separation?)*

---
## Plot 7 — Barplot
**When to use:** Comparing mean of numerical column across categories  
**When NOT to use:** When you need exact counts — use countplot

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Average glucose by outcome
sns.barplot(data=df_clean, x='Outcome', y='Glucose',
            palette=['steelblue', 'salmon'], ax=axes[0])
axes[0].set_title('Average Glucose by Outcome')
axes[0].set_xticklabels(['No Diabetes', 'Diabetes'])
axes[0].set_xlabel('')

# Multiple features mean by outcome
means = df_clean.groupby('Outcome')[
    ['Glucose', 'BMI', 'Age']].mean()
means.T.plot(kind='bar', ax=axes[1],
             color=['steelblue', 'salmon'])
axes[1].set_title('Feature Means by Outcome')
axes[1].set_xticklabels(['Glucose', 'BMI', 'Age'], rotation=0)
axes[1].legend(['No Diabetes', 'Diabetes'])

plt.suptitle('Plot 7: Barplot', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

**Observation:** *(how much higher is average glucose in diabetic patients?)*

---
## Plot 8 — Line Plot
**When to use:** Trends over time or ordered categories  
**When NOT to use:** Unordered categories or two numerical columns — use scatterplot

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Average glucose by age group
df_clean['Age_Group'] = pd.cut(
    df_clean['Age'],
    bins=[20, 30, 40, 50, 60, 70],
    labels=['20s', '30s', '40s', '50s', '60s+']
)

age_glucose = df_clean.groupby('Age_Group')['Glucose'].mean()

axes[0].plot(age_glucose.index, age_glucose.values,
             marker='o', linewidth=2,
             markersize=8, color='steelblue')
axes[0].set_title('Average Glucose by Age Group')
axes[0].set_xlabel('Age Group')
axes[0].set_ylabel('Average Glucose')
axes[0].grid(True, alpha=0.3)

# Diabetes rate by age group
age_diabetes = df_clean.groupby('Age_Group')['Outcome'].mean() * 100

axes[1].plot(age_diabetes.index, age_diabetes.values,
             marker='o', linewidth=2,
             markersize=8, color='salmon')
axes[1].set_title('Diabetes Rate (%) by Age Group')
axes[1].set_xlabel('Age Group')
axes[1].set_ylabel('Diabetes Rate (%)')
axes[1].grid(True, alpha=0.3)

plt.suptitle('Plot 8: Line Plot', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

**Observation:** *(does diabetes rate increase with age? at what age group does it peak?)*

---
## Quick Reference — When to Use Which Plot

| Plot | Use When |
|---|---|
| Histogram | Distribution of one numerical column |
| Boxplot | Outliers + comparing distributions across groups |
| Heatmap | All feature correlations at once |
| Countplot | Category frequencies + class balance |
| Scatterplot | Relationship between two numerical columns |
| Pairplot | All pairwise relationships (max 6 features) |
| Barplot | Mean of numerical column by category |
| Line Plot | Trends over time or ordered categories |
